```python
"""
Add this to your existing notebook/script alongside the table generation code.
This produces a per-seed breakdown for any dataset/score/n/method combination.
"""

import pickle
import numpy as np
import pandas as pd
import glob
import os


def per_seed_results(folder, methods, class_names=None):
    """
    Load all seed pkl files from a folder and return per-seed + averaged results.
    
    Parameters
    ----------
    folder : str
        Path like '.../random_calset/n_totalcal=50/score=APS/'
    methods : dict
        e.g. {'standard': '$\\standard$', 'classwise': '$\\classwise$', ...}
    class_names : list or None
        e.g. ['No DR', 'Mild', 'Moderate', 'Severe', 'Proliferative'] for DDR
        If None, uses Class 0, Class 1, ...
    
    Returns
    -------
    df_per_seed : DataFrame with per-seed metrics (CovGap, AvgSize, per-class coverages)
    df_averaged : DataFrame with mean (SE) across seeds
    """
    # Find all seed files
    seed_files = sorted(glob.glob(os.path.join(folder, 'seed=*_allresults.pkl')))
    n_seeds = len(seed_files)
    
    if n_seeds == 0:
        print(f"No seed files found in {folder}")
        return None, None
    
    # Determine number of classes from first file
    with open(seed_files[0], 'rb') as f:
        d = pickle.load(f)
    first_method = list(methods.keys())[0]
    n_classes = len(d[first_method][2]['raw_class_coverages'])
    
    if class_names is None:
        class_names = [f'Class {i}' for i in range(n_classes)]
    
    # Collect per-seed results
    rows = []
    for seed_file in seed_files:
        seed_num = int(os.path.basename(seed_file).split('=')[1].split('_')[0])

        with open(seed_file, 'rb') as f:
            d = pickle.load(f)
        
        for method_key, method_name in methods.items():
            if method_key not in d:
                continue
            t = d[method_key]
            gap = t[2]['mean_class_cov_gap']
            mcov = t[2]['marginal_cov']
            sz = t[3]['mean']
            cov = t[2]['raw_class_coverages']
            
            row = {
                'Seed': seed_num,
                'Method': method_name,
                'CovGap (%)': round(gap * 100, 1),
                'AvgSize': round(sz, 2),
                'MargCov (%)': round(mcov * 100, 1),
            }
            for i, cname in enumerate(class_names):
                row[cname] = round(cov[i] * 100, 1)
            
            rows.append(row)
    
    df_per_seed = pd.DataFrame(rows)
    
    # Compute averaged results with SE
    avg_rows = []
    for method_key, method_name in methods.items():
        method_df = df_per_seed[df_per_seed['Method'] == method_name]
        if len(method_df) == 0:
            continue
        
        row = {'Method': method_name}
        
        for col in ['CovGap (%)', 'AvgSize', 'MargCov (%)'] + class_names:
            vals = method_df[col].values
            mean = np.mean(vals)
            se = np.std(vals) / np.sqrt(len(vals))
            row[col] = f"{mean:.1f} ({se:.1f})"
        
        # Also find best and worst seed
        gaps = method_df['CovGap (%)'].values
        seeds = method_df['Seed'].values
        row['Best Seed'] = f"Seed {seeds[np.argmin(gaps)]} ({np.min(gaps):.1f}%)"
        row['Worst Seed'] = f"Seed {seeds[np.argmax(gaps)]} ({np.max(gaps):.1f}%)"
        
        avg_rows.append(row)
    
    df_averaged = pd.DataFrame(avg_rows)
    
    return df_per_seed, df_averaged


def print_per_seed_summary(folder, methods, class_names=None, title=None):
    """Pretty-print per-seed results."""
    df_per_seed, df_averaged = per_seed_results(folder, methods, class_names)
    
    if df_per_seed is None:
        return
    
    if title:
        print(f"\n{'='*80}")
        print(f" {title}")
        print(f"{'='*80}")
    
    print(f"\n--- Per-Seed Results ---")
    print(df_per_seed.to_string(index=False))
    
    print(f"\n--- Averaged (Mean ± SE) ---")
    print(df_averaged.to_string(index=False))
    
    return df_per_seed, df_averaged


# ============================================================
# USAGE EXAMPLE
# ============================================================

if __name__ == '__main__':
    
    # Define your methods (same as your existing script)
    methods = {
        'standard': 'standard',
        'classwise': 'classwise',
        'exact_coverage_classwise': 'exact_classwise',
        'exact_coverage_cluster': 'clustered',
        'cluster_random': 'cluster_random'
    }
    
    # DDR class names
    ddr_classes = ['No DR', 'Mild', 'Moderate', 'Severe', 'Proliferative']
    
    # DFUC class names (update these to your actual class names)
    dfuc_classes = ['None', 'Infection', 'Ischaemia', 'Both']
    
    # Base path
    base = r'C:\Anum\0002-Conformal-Pred-Classification\v6\FINAL\conformal\results\paper\varying_n'
    
    # ---- Example: Single config ----
    folder = os.path.join(base, 'ddr', 'ce-ddr', 'random_calset', 'n_totalcal=50', 'score=APS')
    print_per_seed_summary(folder, methods, ddr_classes, title="DDR | CE | n=50 | APS")
    
    folder = os.path.join(base, 'ddr', 'ddr-cp-aware', 'random_calset', 'n_totalcal=50', 'score=APS')
    print_per_seed_summary(folder, methods, ddr_classes, title="DDR | Conformal | n=50 | APS")
    
    # ---- Loop over all configs ----
    datasets = {
        'DDR': {
            'ce': ('ddr', 'ce-ddr', ddr_classes),
            'conf': ('ddr', 'ddr-cp-aware', ddr_classes),
            'focal': ('ddr', 'focal-ddr', ddr_classes),
        },
        'DFUC': {
            'ce': ('dfuc', 'dfuc-ce', dfuc_classes),
            'conf': ('dfuc', 'dfuc-cp-aware', dfuc_classes),
            'focal': ('dfuc', 'focal-dfuc', dfuc_classes),
        },
    }
    scores = ['softmax', 'APS', 'RAPS']
    n_values = [50, 100, 200, 300, 400]
    
    # Collect everything into one big DataFrame
    all_results = []
    for ds_name, losses in datasets.items():
        for loss_name, (ds_dir, loss_dir, cls_names) in losses.items():
            for n in n_values:
                for score in scores:
                    folder = os.path.join(base, ds_dir, loss_dir, 'random_calset', 
                                          f'n_totalcal={n}', f'score={score}')
                    df_per_seed, df_avg = per_seed_results(folder, methods, cls_names)
                    if df_per_seed is not None:
                        df_per_seed['Dataset'] = ds_name
                        df_per_seed['Loss'] = loss_name
                        df_per_seed['n'] = n
                        df_per_seed['Score'] = score
                        all_results.append(df_per_seed)
    
    if all_results:
        df_all = pd.concat(all_results, ignore_index=True)
        
        # Find the best seed for each config
        print("\n\n=== Best seed per config (lowest CovGap) ===")
        idx = df_all.groupby(['Dataset', 'Loss', 'Score', 'n', 'Method'])['CovGap (%)'].idxmin()
        print(df_all.loc[idx][['Dataset', 'Loss', 'Score', 'n', 'Method', 'Seed', 'CovGap (%)']].to_string(index=False))
        
        # Example queries:
        # df_all[(df_all['Dataset']=='DDR') & (df_all['Loss']=='ce') & (df_all['Score']=='APS') & (df_all['n']==50)]
        # df_all[df_all['CovGap (%)'] > 15]  # all configs with gap > 15%
```

```python
import os
import pickle
import pandas as pd
from pathlib import Path

# ── Configuration ──────────────────────────────────────────────────────────
BASE_DIR = Path(
    r"C:\Anum\0002-Conformal-Pred-Classification\v6\FINAL\conformal\results\paper\varying_n\ddr\focal-ddr\random_calset"
)
N_TOTALCALS = [50, 100, 200, 300, 400, 500]
SCORES = ["APS", "softmax", "RAPS"]
OUTPUT_CSV = Path(BASE_DIR) / "all_seed_results.csv"

# ── Collect results ────────────────────────────────────────────────────────
rows = []

for n in N_TOTALCALS:
    for score in SCORES:
        score_dir = BASE_DIR / f"n_totalcal={n}" / f"score={score}"

        if not score_dir.exists():
            print(f"[WARN] Directory not found: {score_dir}")
            continue

        # Find all pkl files in this directory (and subdirs if any)
        pkl_files = sorted(score_dir.glob("**/*.pkl"))

        if not pkl_files:
            print(f"[WARN] No .pkl files in {score_dir}")
            continue

        for pkl_path in pkl_files:
            try:
                with open(pkl_path, "rb") as f:
                    data = pickle.load(f)
            except Exception as e:
                print(f"[ERROR] Could not read {pkl_path}: {e}")
                continue

            # --- Adapt this section to your pkl structure ---
            # If data is a dict, flatten it into the row
            if isinstance(data, dict):
                row = {
                    "n_totalcal": n,
                    "score": score,
                    "file": pkl_path.name,
                }
                row.update(data)
                rows.append(row)

            # If data is a list of dicts (one per seed), iterate
            elif isinstance(data, list):
                for item in data:
                    row = {
                        "n_totalcal": n,
                        "score": score,
                        "file": pkl_path.name,
                    }
                    if isinstance(item, dict):
                        row.update(item)
                    else:
                        row["value"] = item
                    rows.append(row)

            # Fallback: store raw value
            else:
                rows.append({
                    "n_totalcal": n,
                    "score": score,
                    "file": pkl_path.name,
                    "value": data,
                })

# ── Export ──────────────────────────────────────────────────────────────────
df = pd.DataFrame(rows)
df.to_csv(OUTPUT_CSV, index=False)
print(f"\nSaved {len(df)} rows to {OUTPUT_CSV}")
print(df.head(10).to_string())
```

```python
import os
import pickle
import pandas as pd
import numpy as np
from pathlib import Path

# ── Configuration ──────────────────────────────────────────────────────────
BASE_DIR = Path(
    r"C:\Anum\0002-Conformal-Pred-Classification\v6\FINAL\conformal\results\paper\varying_n\ddr\focal-ddr\random_calset"
)
N_TOTALCALS = [50, 100, 200, 300, 400, 500]
SCORES = ["APS", "softmax", "RAPS"]

METHODS = [
    "standard",
    "classwise",
    "exact_coverage_classwise",
    "exact_coverage_cluster",
    "cluster_proportional",
    "cluster_doubledip",
    "cluster_random",
]

# Metrics to extract from the results tuple
# tuple structure: (qhat, None, metrics_dict, setsize_dict)
# metrics_dict keys: mean_class_cov_gap, undercov_gap, overcov_gap, max_gap,
#                    very_undercovered, marginal_cov, raw_class_coverages, ...
# setsize_dict keys: mean, [.25, .5, .75, .9] quantiles

METRICS_TO_EXTRACT = [
    "marginal_cov",
    "mean_class_cov_gap",
    "undercov_gap",
    "overcov_gap",
    "max_gap",
    "very_undercovered",
]
SETSIZE_KEY = "mean"  # mean set size

OUTPUT_CSV = BASE_DIR / "ce_ddr_results_clean.csv"

# ── Helper ─────────────────────────────────────────────────────────────────
def extract_metrics(result_tuple):
    """Extract metrics from a single method's result tuple."""
    # result_tuple = (qhat, None, metrics_dict, setsize_dict)
    if result_tuple is None:
        return {}

    try:
        metrics_dict = result_tuple[2]  # 3rd element
        setsize_dict = result_tuple[3]  # 4th element
    except (IndexError, TypeError):
        return {}

    row = {}
    for key in METRICS_TO_EXTRACT:
        val = metrics_dict.get(key, np.nan)
        row[key] = float(val) if val is not None else np.nan

    # Mean set size
    if setsize_dict and SETSIZE_KEY in setsize_dict:
        row["mean_setsize"] = float(setsize_dict[SETSIZE_KEY])
    else:
        row["mean_setsize"] = np.nan

    # Set size quantiles
    quantile_key = "[.25, .5, .75, .9] quantiles"
    if setsize_dict and quantile_key in setsize_dict:
        q = setsize_dict[quantile_key]
        for i, label in enumerate(["q25", "q50", "q75", "q90"]):
            row[f"setsize_{label}"] = float(q[i]) if i < len(q) else np.nan

    return row


# ── Collect results ────────────────────────────────────────────────────────
rows = []

for n in N_TOTALCALS:
    for score in SCORES:
        score_dir = BASE_DIR / f"n_totalcal={n}" / f"score={score}"

        if not score_dir.exists():
            print(f"[WARN] Missing: {score_dir}")
            continue

        pkl_files = sorted(score_dir.glob("*.pkl"))
        if not pkl_files:
            print(f"[WARN] No pkl in: {score_dir}")
            continue

        for pkl_path in pkl_files:
            # Extract seed number from filename like "seed=0_allresults.pkl"
            fname = pkl_path.stem  # e.g. "seed=0_allresults"
            try:
                seed = int(fname.split("seed=")[1].split("_")[0])
            except (IndexError, ValueError):
                seed = fname

            try:
                with open(pkl_path, "rb") as f:
                    data = pickle.load(f)
            except Exception as e:
                print(f"[ERROR] {pkl_path}: {e}")
                continue

            for method in METHODS:
                if method not in data:
                    continue

                metrics = extract_metrics(data[method])

                row = {
                    "n_totalcal": n,
                    "score": score,
                    "seed": seed,
                    "method": method,
                }
                row.update(metrics)
                rows.append(row)

# ── Build DataFrame & save ─────────────────────────────────────────────────
df = pd.DataFrame(rows)

# Reorder columns
id_cols = ["n_totalcal", "score", "seed", "method"]
metric_cols = [c for c in df.columns if c not in id_cols]
df = df[id_cols + sorted(metric_cols)]

df.to_csv(OUTPUT_CSV, index=False)
print(f"\nSaved {len(df)} rows  →  {OUTPUT_CSV}")
print(f"\nShape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"\nn_totalcal values: {sorted(df['n_totalcal'].unique())}")
print(f"Scores: {sorted(df['score'].unique())}")
print(f"Seeds: {sorted(df['seed'].unique())}")
print(f"Methods: {sorted(df['method'].unique())}")
print(f"\nFirst 10 rows:")
print(df.head(10).to_string(index=False))
```
